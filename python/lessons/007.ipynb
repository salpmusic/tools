# Python入門 第7回：while文の基本 ～BASICのWHILE〜WENDと比べてみよう～

```python
# 第7回 while文の基本
# BASICのWHILE〜WENDと比べよう
print("while文の練習")

# 1から5まで数える
count = 1
while count <= 5:
    print("カウント:", count)
    count = count + 1
print("カウント終了！")

# 10から2ずつ減らす
number = 10
while number > 0:
    print("残り:", number)
    number = number - 2
print("すべて終了しました")
```
